## Task 1

```python
transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]
```

```python
# Filter out invalid transactions
valid_transactions = []

for t in transactions:
    if t["price"] > 0 and t["quantity"] >= 0:
        valid_transactions.append(t)

valid_transactions
```

```python
# Add total revenue for each valid transaction
for t in valid_transactions:
    t["total"] = t["price"] * t["quantity"]

valid_transactions
```

```python
revenue_per_region = {}

for t in valid_transactions:
    region = t["region"]
    
    if region not in revenue_per_region:
        revenue_per_region[region] = 0
    
    revenue_per_region[region] += t["total"]

revenue_per_region
```

```python
# Find the region with highest revenue
best_region = max(revenue_per_region, key=revenue_per_region.get)
best_value = revenue_per_region[best_region]

print(f"Best region: {best_region} with revenue {best_value}")
```

## Task 2

```python
sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]
```

```python
def summarise_sales(sales):
    # Calculate total and average sales
    total = sum(sales)
    average = round(total / len(sales), 2)

    # Find best and worst day ( starting from day 1 )
    best_day = sales.index(max(sales)) + 1
    worst_day = sales.index(min(sales)) + 1
    
    # Count how many days were above average
    above_average_count = 0
    for value in sales:
        if value > average:
            above_average_count += 1

    return {
        "total": total,
        "average": average,
        "best_day": best_day,
        "worst_day": worst_day,
        "above_average_count": above_average_count
    }
```

```python
result = summarise_sales(sales)
print(result)
```

## Task 4

```python
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.holdings = {} # Store asset by ticker symbol

    def add_asset(self, ticker, shares, price_per_share):
        # Add a new asset or update an existing one
        self.holdings[ticker] = {
            "shares": shares,
            "price_per_share": price_per_share
        }

    def total_value(self):
        # Calculate the total market value of the portfolio
        total = 0
        for asset in self.holdings.values():
            total += asset["shares"] * asset["price_per_share"]
        return total

    def most_valuable(self):
        # Find the asset with the highest total value
        best = None
        max_value = 0

        for ticker, asset in self.holdings.items():
            value = asset["shares"] * asset["price_per_share"]
            if value > max_value:
                max_value = value
                best = ticker

        return best

    def __str__(self):
        return f"Owner: {self.owner}, Assets: {len(self.holdings)}, Total Value: {self.total_value()}"
```

```python
p = Portfolio("Alex")

p.add_asset("AAPL", 10, 150)
p.add_asset("TSLA", 5, 700)
p.add_asset("GOOG", 2, 2000)

print(p)
print("Total value:", p.total_value())
print("Most valuable:", p.most_valuable())
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=123)
days = pd.date_range(start="2025-01-06", periods=28, freq="D")
visitors = rng.integers(200, 600, size=28)
conversion_rate = rng.uniform(0.02, 0.08, size=28).round(4)
avg_order_value = rng.normal(50, 10, size=28).round(2)
```

### Task 8a

```python
# Create a DataFrame and set date as the index
df = pd.DataFrame({
    "date": days,
    "visitors": visitors,
    "conversion_rate": conversion_rate,
    "avg_order_value": avg_order_value
})

df = df.set_index("date")

df
```

### Task 8b

```python
# Calculate daily conversions and revenue
df["conversions"] = (df["visitors"] * df["conversion_rate"]).round()

df["daily_revenue"] = df["conversions"] * df["avg_order_value"]

df
```
